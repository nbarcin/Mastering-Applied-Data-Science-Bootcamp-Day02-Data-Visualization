```python
## Bar Chart
# Use the following data to make your plot:

from numpy import arange, array
x = arange(1, 13)
y = array([7, 8, 5, 5, 3, 4, 4, 6, 9, 7, 8, 8])
```

1. Add a bar chart to your plot, using `x` and `y` above. Fix it so there is no outline (edgecolor), set the color of the bars to green, and set their width to 0.25.
2. Fix your X-axis to go from 1 to 13.
3. Fix your Y-axis to go from 0 to 10.
4. Add a label to your X-axis that says "month of the year".
5. Add a label to your Y-axis that says "letters in month name".
6. Add a title to your plot, "Leters in the Name of Each Month".
7. Display your plot.

### Bar Chart

```python
# Solution
```

```python
# Import NumPy for numerical operations
import numpy as np

# Create an array of integers from 1 to 12 (representing the months)
x = np.arange(1, 13)

# Create an array containing the number of letters in each month's name
y = np.array([7, 8, 5, 5, 3, 4, 4, 6, 9, 7, 8, 8])

# Import pyplot from Matplotlib for creating visualizations
from matplotlib import pyplot as plt

# Create a bar chart with green bars, a width of 0.25, and no edge lines
plt.bar(x, y, width=0.25, color='green', edgecolor='none')

# Set the x-axis limits from 1 to 13
plt.xlim(1, 13)

# Set the y-axis limits from 0 to 10
plt.ylim(0, 10)

# Add a label to the x-axis
plt.xlabel('Month of the Year')

# Add a label to the y-axis
plt.ylabel('Letters in Month Name')

# Set the chart title
plt.title('Letters in the Name of Each Month')

# Display the bar chart
plt.show()
```

### Summary

* **Data Creation:** Uses NumPy to create an array representing the 12 months of the year and the number of letters in each month's name.
* **Bar Chart:** Uses Matplotlib's `plt.bar()` to visualize the number of letters in each month.
* **Plot Customization:** Adjusts bar width, color, axis limits, labels, and chart title.
* **Data Visualization:** Demonstrates how to present categorical data using a simple bar chart.

### Scatter and Line Plots

```python
# Solution
```

```python
import numpy as np
from matplotlib import pyplot as plt

# Create x values from 0 to 2*pi
x = np.arange(0.0, 2 * np.pi, np.pi / 16.0)

# Calculate y values using the cosine function
y = 2 * np.cos(x)

# Draw a dashed green line
plt.plot(x, y, linewidth=3, c='green',
         linestyle='--', label='approx')

# Display the data points as blue dots
plt.scatter(x, y, s=300, c='blue',
            edgecolor='none', label='approx')

# Add axis labels and title
plt.xlabel('x')
plt.ylabel('y')
plt.title('2 * cos(x)')

# Display the legend
plt.legend()

# Save the graph
plt.savefig('cos.png', bbox_inches='tight')

# Display the graph
plt.show()
```

### Summary

* **Data Generation:** Uses NumPy to generate x-values from 0 to \(2\pi\) and calculates y-values using the cosine function, \(y = 2\cos(x)\).
* **Line Plot:** Draws a dashed green line to represent the cosine curve.
* **Scatter Plot:** Displays the data points as large blue dots over the line.
* **Plot Customization:** Adds axis labels, a title, and a legend to make the graph easier to understand.
* **Saving the Figure:** Saves the visualization as `cos.png` using `plt.savefig()`.

### Multiplot, Histogram, and Errorbar

```python
# Solution
```

```python
# Import NumPy for numerical operations and random number generation
import numpy as np

# Define the mean (average) of the normal distribution
mu = 10500

# Define the standard deviation, which measures the spread of the data
sigma = 250

# Generate 10,000 random values from a normal distribution
values = mu + sigma * np.random.randn(10000)

# Generate 100 random x-values from the same normal distribution
x = mu + sigma * np.random.randn(100)

# Generate 100 random y-values from the same normal distribution
y = mu + sigma * np.random.randn(100)

# Import pyplot from Matplotlib for creating visualizations
from matplotlib import pyplot as plt

# Create the first subplot in a layout with 1 row and 2 columns
plt.subplot(1, 2, 1)

# Create a histogram with 25 bins and green bars
plt.hist(values, 25, facecolor='green', edgecolor='none')

# Label the y-axis of the histogram
plt.ylabel('Count')

# Create the second subplot in the same 1-row, 2-column layout
plt.subplot(1, 2, 2)

# Create a scatter plot with semi-transparent, large points in random colors
plt.scatter(x, y, s=300, alpha=0.25, edgecolor='none',
            c=np.random.rand(100))

# Label the x-axis of the scatter plot
plt.xlabel('X')

# Label the y-axis of the scatter plot
plt.ylabel('Y')

# Set the x-axis range from 10,000 to 11,000
plt.xlim(10000, 11000)

# Set the y-axis range from 10,000 to 11,000
plt.ylim(10000, 11000)

# Display grid lines on the current subplot
plt.grid()

# Display both plots
plt.show()
```

### Summary

* **Histogram:** Displays the distribution of 10,000 randomly generated values with a mean of 10,500 and a standard deviation of 250.
* **Scatter Plot:** Visualizes 100 randomly generated `(x, y)` pairs using different colors and semi-transparent points.
* **Subplots:** Uses `plt.subplot()` to display two visualizations side by side.
* **Customization:** Demonstrates how to adjust axis limits, point size, transparency, colors, and grid lines using Matplotlib.
