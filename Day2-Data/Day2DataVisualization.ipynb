### Radar

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temperature = [10, 12, 15, 18, 21, 24, 27, 26, 23, 19, 14, 11]

# Convert data to a dataframe
df = pd.DataFrame({'Temperature': temperature}, index=months)

# Create the polar chart using the dataframe
fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})
ax.plot(np.radians(np.arange(0, 360, 30)), df['Temperature'])
ax.fill(np.radians(np.arange(0, 360, 30)), df['Temperature'], alpha=0.5)
ax.set_xticks(np.radians(np.arange(0, 360, 30)))
ax.set_xticklabels(months);
```

### Sankey diagram

```python
import plotly.graph_objects as go

labels = ["Source A", "Source B", "Source C", "Target X", "Target Y", "Target Z"]
source = [0, 1, 0, 2, 3, 3, 4]
target = [3, 3, 4, 4, 5, 5]
values = [8, 4, 2, 8, 4, 2, 3]

fig = go.Figure(data=[go.Sankey(
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=labels
    ),
    link=dict(
        source=source,
        target=target,
        value=values
    )
)])

fig.update_layout(title_text="Basic Sankey Diagram", font_size=10)
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Sample data
X = np.linspace(-5, 5, 100)
Y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create a 3D surface plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')

plt.title('3D Surface Plot')
plt.show()
```
